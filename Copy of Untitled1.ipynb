```python
from google.colab import drive
import pandas as pd
```

```python
df = pd.read_csv('/content/spotify_tracks (1).csv')

df.head()
```

```python
df.tail()
```

```python
print("Available columns:", df.columns.tolist())
```

```python
for col in df.columns: df[col] = pd.to_numeric(df[col], errors='ignore')
```

```python
numeric_features = [ 'popularity','danceability','energy','tempo', 'loudness','duration_ms','acousticness','valence' ]
numeric_features = [f for f in numeric_features if f in df.columns]
print("Numeric features to plot:", numeric_features)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for col in numeric_features:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col].dropna(), bins=40, kde=True, color="skyblue")
    plt.title(f"Histogram of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

for feat in numeric_features:
  if feat in df.columns:
     plt.figure(figsize=(6,4))
     sns.histplot(df[feat].dropna(), bins=40, kde=True, color="skyblue")
     plt.title(f"Histogram of {feat}")
     plt.xlabel(feat)
     plt.ylabel("Count")
     plt.show()
```

```python
cat_col = 'artists' if 'artists' in df.columns else df.select_dtypes(exclude=['number']).columns[0]
print("Categorical column chosen for donut:", cat_col)
# Top 5 categories
top5 = df[cat_col].value_counts().nlargest(5)
```

```python
plt.figure(figsize=(6,6))
plt.pie(top5, labels=top5.index, autopct='%1.1f%%', startangle=140)
centre_circle = plt.Circle((0,0), 0.70, fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.title(f"Top 5 {cat_col} Distribution (Donut Chart)")
plt.show()
```

```python
# Use release_date if available, else index
if 'release_date' in df.columns:
    df['release_date'] = pd.to_datetime(df['release_date'], errors='coerce')
    df_sorted = df.sort_values('release_date')
    x = df_sorted['release_date']
    y = df_sorted['popularity'] if 'popularity' in df.columns else df_sorted.select_dtypes(include=['number']).iloc[:,0]
    ylabel = 'popularity' if 'popularity' in df.columns else y.name
else:
    x = df.index
    y = df['popularity'] if 'popularity' in df.columns else df.select_dtypes(include=['number']).iloc[:,0]
    ylabel = 'popularity' if 'popularity' in df.columns else y.name

plt.figure(figsize=(10,5))
plt.plot(x, y, marker='o', linestyle='-', color='green')
plt.title(f"Line Chart of {ylabel} over time/index")
plt.xlabel("Release Date" if 'release_date' in df.columns else "Index")
plt.ylabel(ylabel)
plt.xticks(rotation=45)
plt.show()
```
